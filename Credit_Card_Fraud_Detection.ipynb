```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve, auc
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
import numpy as np
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)


best_models = {}
```

```python
from sklearn.linear_model import LogisticRegression

lr_params = {
    "C": [0.001, 0.01, 0.1, 1, 10],
    "penalty": ["l1", "l2"],
    "solver": ["liblinear", "saga"],
    "max_iter": [100,200,300]
}

lr_search = RandomizedSearchCV(
    LogisticRegression(),
    lr_params,
    n_iter=15,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

lr_search.fit(X_train_res, y_train_res)

print("Best LR Params:", lr_search.best_params_)
```

```python
lr_best = lr_search.best_estimator_
best_models["Logistic Regression"] = lr_best
```

```python
y_pred_lr = lr_best.predict(X_test)
y_proba_lr = lr_best.predict_proba(X_test)[:,1]

print("\n====== Logistic Regression ========")
print(classification_report(y_test, y_pred_lr))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_lr))

prec_lr = precision_score(y_test, y_pred_lr)
rec_lr = recall_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr)

prec_, rec_, _ = precision_recall_curve(y_test, y_proba_lr)
pr_auc_lr = auc(rec_, prec_)

print(f"Precision: {prec_lr:.4f}, Recall: {rec_lr:.4f}, F1: {f1_lr:.4f}, PR AUC: {pr_auc_lr:.4f}")
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_params = {
    "criterion": ["gini", "entropy"],
    "max_depth": [None, 5, 10, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "class_weight": [None, "balanced"]
}

dt_search = RandomizedSearchCV(
    DecisionTreeClassifier(random_state=42),
    dt_params,
    n_iter=20,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)
dt_search.fit(X_train_res, y_train_res)
print("Best DT Params:", dt_search.best_params_)
```

```python
dt_best = dt_search.best_estimator_
best_models["Decision Tree"] = dt_best
```

```python
y_pred_dt = dt_best.predict(X_test)
print("\n==== Decision Tree ====")
print(classification_report(y_test, y_pred_dt))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_dt))
prec_dt = precision_score(y_test, y_pred_dt)
rec_dt = recall_score(y_test, y_pred_dt)
f1_dt = f1_score(y_test, y_pred_dt)
print(f"Precision: {prec_dt:.4f}, Recall: {rec_dt:.4f}, F1: {f1_dt:.4f}")
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_params = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "bootstrap": [True, False],
    "class_weight": [None, "balanced"]
}
rf_search = RandomizedSearchCV(
    RandomForestClassifier(random_state=42),
    rf_params,
    n_iter=20,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)
rf_search.fit(X_train_res, y_train_res)
print("Best RF Params:", rf_search.best_params_)
```

```python
rf_best = rf_search.best_estimator_
best_models["Random Forest"] = rf_best
```

```python
y_pred_rf = rf_best.predict(X_test)
y_proba_rf = rf_best.predict_proba(X_test)[:,1]
print("\n =====Random Forest ====")
print(classification_report(y_test, y_pred_rf))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_rf))
prec_rf = precision_score(y_test, y_pred_rf)
rec_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)
prec_, rec_, _ = precision_recall_curve(y_test, y_proba_rf)
pr_auc_rf = auc(rec_, prec_)
print(f"Precision: {prec_rf:.4f}, Recall: {rec_rf:.4f}, F1: {f1_rf:.4f}, PR AUC: {pr_auc_rf:.4f}")
```

```python
results = []
for name, model in best_models.items():
    y_pred = model.predict(X_test)
    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:,1]
        prec_, rec_, _ = precision_recall_curve(y_test, y_proba)
        pr_auc = auc(rec_, prec_)
    else:
        pr_auc = None

    results.append({
        "Model": name,
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred),
        "PR AUC": pr_auc
    })

import pandas as pd
df_results = pd.DataFrame(results)
print(df_results.sort_values("F1 Score", ascending=False))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve, auc

plt.figure(figsize=(8,6))

for name, model in best_models.items():
    
    if hasattr(model, "predict_proba"):
        scores = model.predict_proba(X_test)[:, 1]
    elif hasattr(model, "decision_function"):
        scores = model.decision_function(X_test)
    else:
        print(f"Skipping {name} — no probability or decision_function available.")
        continue

    prec, rec, _ = precision_recall_curve(y_test, scores)
    pr_auc = auc(rec, prec)   
    plt.plot(rec, prec, label=f"{name} (PR AUC={pr_auc:.3f})")

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve — All Models")
plt.legend(loc="best")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

plt.figure(figsize=(8,6))

for name, model in best_models.items():
    if hasattr(model, "predict_proba"):
        scores = model.predict_proba(X_test)[:, 1]
    elif hasattr(model, "decision_function"):
        scores = model.decision_function(X_test)
    else:
        print(f"Skipping {name} — no probability or decision_function available.")
        continue

    fpr, tpr, _ = roc_curve(y_test, scores)
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f"{name} (ROC AUC={roc_auc:.3f})")

# diagonal line for random classifier
plt.plot([0,1], [0,1], linestyle='--', linewidth=1)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve — All Models")
plt.legend(loc="best")
plt.grid(True)
plt.tight_layout()
plt.show()
```
